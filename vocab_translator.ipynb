```python
#!pip install deep-translator
from deep_translator import GoogleTranslator
import json
import time

# C2 Seviyesi (İleri Düzey) İngilizce Kelime Listesi
ingilizce_kelimeler = [
    "ephemeral",
    "sycophant",
    "esoteric",
    "fastidious",
    "obfuscate",
    "mellifluous",
    "recalcitrant",
    "perfunctory",
    "equivocal",
    "ineffable",
    "paradigm",
    "serendipity",
    "vicarious",
    "zealous",
    "anomaly",
    "loquacious"
]

ceviri_sozlugu = {}

translator = GoogleTranslator(source='en', target='tr')

print(f"Toplam {len(ingilizce_kelimeler)} adet C2 seviye kelime çevriliyor. Lütfen bekleyin...\n")

for kelime in ingilizce_kelimeler:
    try:
        turkcesi = translator.translate(kelime)
        ceviri_sozlugu[kelime] = turkcesi

        time.sleep(0.2)
    except Exception as e:
        print(f"Hata! '{kelime}' çevrilemedi: {e}")

cikti = json.dumps(ceviri_sozlugu, ensure_ascii=False, indent=4)
print(cikti)
```
